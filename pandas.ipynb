```python
# ex) 날짜별로 최고의 가격을 보고 싶다!!
#     (단, 고가의 컬럼이외에 다른 컬럼에도 최고값이 있다고 가정!!!!!!)
# --> 체크 : 일단 하고자 하는 일이 줄 단위로 동일한지 : OK
#     기능/규칙 : 임의의 가로줄에서,,,여러개 값 중에서 최고값 찾자!!
# ==> apply( 모든 줄에 있는 값을 롤링 : for)
#     + lambda( 할 일 / 규칙 서술!!!)
```

```python
stock_price_df.apply( lambda x : x.max(), axis=0)
```

```python
# 날짜별로 그 날의 변동성을 체크!!!
# ==> 그 날의 max -  그 날의 min
# ==> 체크 : 줄 단위(가로 - 날짜 줄 별로), 할 일이 규칙 OK
#     규칙 : 그 날의 max - 그 날의 min
```

```python
stock_price_df.apply( lambda x : x.max()-x.min(), axis=1)
```

```python
stock_price_df["Gap"]=stock_price_df.apply( lambda x : x.max()-x.min(), axis=1)
stock_price_df
```

```python
stock_price_df["GapFor"] = 0
stock_price_df
```

```python
# for x in range(len(
stock_price_df["GapFor"]=stock_price_df["고가"]-stock_price_df["저가"]
stock_price_df
```

```python
gap_list = []

for idx in stock_price_df.index:
    high = stock_price_df.loc[idx, "고가"]
    low = stock_price_df.loc[idx, "저가"]

    gap_list.append(high - low)

stock_price_df["GapFor"] = gap_list
stock_price_df
```
